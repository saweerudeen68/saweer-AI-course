```python

# Travel Agent
# A simple rule-based intelligent agent

print("Welcome to Smart Travel Agent!")

budget = int(input("Enter your travel budget in PKR: "))
interest = input(
    "Choose your interest (mountains, beach, history): "
).lower()

if budget < 10000:
    print("Recommendation: Explore a nearby local destination.")

elif interest == "mountains":
    if budget >= 30000:
        print("Recommended destination: Hunza, Gilgit-Baltistan")
    else:
        print("Recommended destination: Murree, Pakistan")

elif interest == "beach":
    print("Recommended destination: Karachi Beach, Pakistan")

elif interest == "history":
    print("Recommended destination: Lahore, Pakistan")

else:
    print("Please choose mountains, beach, or history.")

print("Thank you for using Smart Travel Agent!")
```
